```python
import pandas as pd
import numpy as np
import os
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
pd.options.plotting.backend = "plotly"
import joblib
```

```python
# reads in a single text file, assigns them column names, and adds the label (alphabet) column
def with_labels(fp, letter):
    l = []
    with open(fp) as f:
        for line in f:
            # first line contains PuTTy garbage
            if line[0] == '=':
                continue
            l.append(line.strip().split(','))

    df = pd.DataFrame(l)#.drop(columns = [0])
    df.columns = ['zeroth', 'thumb', 'point', 'middle', 'ring', 'pinky']
    # df.columns = ['thumb', 'point', 'middle', 'ring', 'pinky']
    df = df.astype(int)
    df['label'] = letter
    return df

# creates two dataframes from all text files in folder_fp
# uses above function to convert text file to a dataframe, then splits them into training and testing,
# returns training and testing dataframe
def make_dataset(folder_fp):
    csvs = os.listdir(folder_fp)
    train = []
    test = []
    for csv in csvs:
        # exclude h and g from training
        if 'txt' in csv and 'h' not in csv and 'g' not in csv:
            df = with_labels(os.path.join(folder_fp, csv), csv[0])
            df_train, df_test = train_test_split(df, test_size=0.2, random_state = True)
            train.append(df_train)
            test.append(df_test)
    train = pd.concat(train)
    test = pd.concat(test)
    train = train.reset_index().drop(columns = ['index'])
    test = test.reset_index().drop(columns = ['index'])
    return train, test
```

```python
# fit = training, score = accuracy of the model
train, test = make_dataset('../csv_files')

# # you can try changing n_neighbors to see how model performance is affected
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(train[['zeroth', 'thumb', 'point', 'middle', 'ring', 'pinky']], train['label'])
knn.score(test[['zeroth', 'thumb', 'point', 'middle', 'ring', 'pinky']], test['label'])

fn = 'knn_model_5.sav'
joblib.dump(knn, fn)
```

```python
test['predicted'] = test.apply(lambda x: knn.predict([x[['zeroth', 'thumb', 'point', 'middle', 'ring', 'pinky']]])[0], axis = 1)
test
```

```python
test['(actual, pred)'] = test.apply(lambda x: (x['label'], x['predicted']), axis = 1)
test
```

```python
test[test['label'] != test['predicted']]['(actual, pred)'].value_counts()
```

```python
# plotting accuracy for each letter

score_df = pd.DataFrame(knn.predict(test[['thumb', 'point', 'middle', 'ring', 'pinky']]))
score_df.columns = ['predicted']
score_df = score_df.merge(test['label'], left_index = True, right_index=True)
score_df

score_df['correct'] = score_df['label'] == score_df['predicted']
pd.DataFrame(score_df.groupby('label')['correct'].mean()).plot(kind = 'bar')
```

```python
# finding the pairs of alphabets that the model got most wrong

tup = score_df[score_df['predicted'] != score_df['label']].apply(lambda row: tuple([row['label'], row['predicted']]), axis = 1)
# pd.DataFrame(tup).reset_index().set_index(0).plot(kind = 'bar')
tup_df = pd.DataFrame(tup.value_counts()).reset_index()

tup_df['index'] = tup_df['index'].astype(str)
fig = tup_df.iloc[:10].plot(kind = 'bar', x = 'index', y = 'count')
fig.update_layout(title = 'Pairs of Alphabet That The Model Got Most Wrong (Actual, Predicted)')
```
